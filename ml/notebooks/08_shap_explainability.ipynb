```python
import sys
from pathlib import Path
import pandas as pd

# Connect to src directory
ml_root = Path().resolve().parents[0]
sys.path.append(str(ml_root))

from src.explain import ModelExplainer

# Load Processed Data
train = pd.read_csv(ml_root / "data" / "processed" / "train_FD001_features.csv")

# Initialize Explainer
model_path = ml_root.parent / "models" / "xgb_rul_model.pkl"
explainer = ModelExplainer(model_path)

# Grab the feature columns automatically extracted by our explainer
feature_cols = explainer.feature_cols
X = train[feature_cols]

print("SHAP Explainer Initialized Successfully.")
```

```python
# This takes a minute as it calculates SHAP values for the entire dataset
explainer.summary_plot(X)
print("Summary Plot saved to docs/shap_summary.png")
```

```python
# Find an engine on the brink of failure
near_failure_idx = train["RUL"].idxmin()

prediction, contributions = explainer.explain_prediction(X, near_failure_idx)

print(f"Predicted Remaining Useful Life : {prediction:.2f} Flights")
print("\nTop Contributing Factors:")
display(contributions.head(10))

# Generate Local Plots
explainer.waterfall_plot(X, near_failure_idx)
explainer.force_plot(X, near_failure_idx)
print("\nLocal Explanation Plots saved to docs/")
```
